[변수들에 대한 설명]

Age : 나이

Gender : 성별

Height (m) : 신장

Weight (kg) : 체중

BMI : 체질량지수

Fat_Percentage : 체지방률

Workout_Type : 운동 종류

Workout_Frequency (days/week) : 주당 운동 횟수

Session_Duration (hours) : 세션 지속 시간

Experience_Level : 숙련도

Resting_BPM : 안정시 심박수

Avg_BPM : 평균 심박수

Max_BPM : 최대 심박수

Calories_Burned : 소모 칼로리

Water_Intake (liters) : 수분 섭취량

A유형: 🧚 유산소 요정 (Cardio Fairy)

데이터 특징: Workout_Type = Cardio/HIIT, Avg_BPM 140 이상 높게 유지.

설명: 런닝머신과 천국의 계단을 지배하는 자들입니다. 웨이트 존의 거친 호흡을 피해 유산소 존에서 묵묵히 칼로리를 태웁니다. 이들의 심장은 쉴 새 없이 뛰며, 세션 내내 높은 평균 심박수를 유지해 가장 훌륭한 체지방 커팅 능력을 자랑합니다.

B유형: 🦍 4분할 쇳질 장인 (Iron Craftsman)

데이터 특징: Workout_Type = Strength, Workout_Frequency = 4회, Max_BPM 간헐적 폭발.

설명: 가슴, 등, 어깨, 하체로 이어지는 정석적인 4일 주기 분할 루틴을 신앙처럼 지킵니다. 가볍게 40kg 수준의 중량으로 웜업을 시작해 와이드 체스트 프레스나 V-스쿼트 같은 고중량 본 세트에 진입할 때 심박수가 극한으로 폭발하는 패턴을 보입니다. 단단하고 넓은 프레임을 목표로 가장 체계적이고 이상적인 근성장 테크트리를 타고 있습니다.

C유형: 💸 헬스장 기부 천사 (Donation Angel)

데이터 특징: Session_Duration 0.5시간 미만, Workout_Frequency 1~2회.

설명: 헬스장 관장님들이 가장 사랑하는 VIP 고객입니다. 야심 차게 등록한 후 가끔 나타나 폼롤러만 굴리다 집에 가거나, 아예 나타나지 않습니다. 운동 시간 대비 가장 비싼 비용을 지불하고 있는 이타적인 영혼들입니다.

D유형: 🚜 오버트레이닝 불도저 (Overtraining Bulldozer)

데이터 특징: Workout_Frequency 6~7회, Session_Duration 2시간 이상.

설명: 헬스장을 제2의 거주지로 삼은 사람들입니다. 하루라도 쇠를 들지 않으면 근손실이 올 거라는 강박에 시달립니다. 높은 숙련도를 자랑하지만, 근육이 회복될 틈을 주지 않아 피로도 누적으로 인한 부상 위험도가 가장 높은 아슬아슬한 집단입니다.
